## 1. Scientific Background on SKR

Saturn Kilometric Radiation (SKR) is a non-thermal radio emission originating from Saturn's auroral regions. It was first detected by the Voyager spacecraft and subsequently studied in detail by the Cassini mission. Key characteristics include:

- **Frequency range**: Typically observed between 3 kHz and 1.2 MHz.
- **Source region**: Generated via the Cyclotron Maser Instability (CMI) in Saturn's high-latitude magnetosphere.
- **Modulation**: SKR intensity is modulated by Saturn's rotation and is influenced by solar wind conditions.
- **Seasonal variation**: The northern and southern SKR sources exhibit different rotation rates, linked to seasonal changes in Saturn's ionosphere.

The interaction between the solar wind and Saturn's magnetosphere plays a critical role in modulating SKR emission. Increased solar wind dynamic pressure compresses the magnetosphere, enhancing auroral currents and consequently SKR intensity.
