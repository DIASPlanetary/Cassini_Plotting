## 1. Scientific Background on SKR

Saturn Kilometric Radiation (SKR) is a non-thermal radio emission from Saturn's auroral regions, first detected by the Voyager spacecraft in 1980. SKR is generated by the cyclotron maser instability and is closely associated with Saturn's auroral processes.

Key characteristics of SKR:
- Frequency range: ~3 kHz to ~1.2 MHz
- Source region: Saturn's high-latitude ionosphere/magnetosphere
- Modulation: SKR intensity is modulated by Saturn's rotation (period ~10.7 hours)
- Solar wind dependence: SKR intensity increases with solar wind dynamic pressure

The Cassini Radio and Plasma Wave Science (RPWS) instrument provided extensive observations of SKR throughout the mission (2004-2017).
