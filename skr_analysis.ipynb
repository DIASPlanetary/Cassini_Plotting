## 1. Scientific Background on SKR

Saturn Kilometric Radiation (SKR) is a non-thermal radio emission from Saturn's auroral regions, primarily observed at frequencies between a few kHz and several hundred kHz. SKR is generated by the cyclotron maser instability and is closely linked to auroral processes at Saturn.

Key characteristics of SKR:
- **Frequency range**: Typically observed between 3 kHz and 1.2 MHz
- **Periodicity**: Exhibits a planetary period modulation (~10.7 hours) that differs between northern and southern hemispheres
- **Solar wind correlation**: SKR intensity is strongly correlated with solar wind dynamic pressure, suggesting a driven magnetospheric system
- **Morphology**: Consists of narrow-band emissions with both left-hand and right-hand circular polarization
