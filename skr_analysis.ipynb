## 1. Scientific Background on SKR

Saturn Kilometric Radiation (SKR) is a non-thermal radio emission from Saturn's auroral regions, discovered by the Voyager spacecraft in 1980. SKR is generated by the Cyclotron Maser Instability (CMI) mechanism, driven by energetic electrons precipitating along magnetic field lines into Saturn's polar ionosphere.

Key properties of SKR:
- **Frequency range**: Typically 3 kHz to 1.2 MHz
- **Source region**: High-latitude auroral zones of Saturn
- **Polarization**: Primarily Right-Hand Circular (RHC) in the Northern Hemisphere, Left-Hand Circular (LHC) in the Southern Hemisphere
- **Variability**: Strongly modulated by solar wind dynamic pressure and Saturn's rotation period

The intensity of SKR is known to be strongly correlated with solar wind conditions. When the solar wind dynamic pressure increases (e.g., during a solar wind compression event), the magnetospheric field lines are compressed, enhancing particle precipitation and thus increasing SKR emission.

## 2. Dataset Description: Cassini RPWS Data (2004)

The dataset used in this module comes from the **Cassini Radio and Plasma Wave Science (RPWS)** instrument during the year 2004, when Cassini was approaching Saturn.

### Data Parameters:
- **SKR Spectral Power Density**: Measured by the RPWS High Frequency Receiver (HFR), in units of W/m²/Hz
- **Frequency channels**: 3.95 kHz to 1.158 MHz (select channels from the HFR)
- **Time resolution**: Typically 32 seconds (low rate) to 2 seconds (high rate)
- **Solar wind dynamic pressure**: Derived from Cassini MIMI/CHEMS and CAPS instruments, or interpolated from 1 AU OMNI data propagated to Saturn

### Data Sources:
- RPWS data: Planetary Data System (PDS) via the Cassini RPWS archive
- Solar wind data: OMNI dataset propagated to Saturn's orbit (~9.5 AU)
