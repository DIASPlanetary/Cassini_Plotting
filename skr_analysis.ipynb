## 1. Scientific Background on SKR

Saturn Kilometric Radiation (SKR) is a non-thermal radio emission from Saturn's auroral regions, first detected by the Voyager spacecraft. SKR is generated by the cyclotron maser instability and is closely linked to auroral processes at Saturn. The intensity of SKR is strongly modulated by the solar wind, with increases in solar wind dynamic pressure leading to enhanced SKR emission.
